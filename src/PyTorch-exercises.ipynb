```python
import torch as t

t.manual_seed(0)
tensor1 = t.rand(7,7)
tensor2 = t.rand(1,7)
print(tensor1)
print(tensor2.T)

print(tensor1.matmul(tensor2.T))

t.cuda.manual_seed(0)
#exercises 2-5
```

```python
t.manual_seed(1234)
tensor3 = t.rand(2,3)
tensor4 = t.rand(2,3)

matmulTensor = t.matmul(tensor3, tensor4.T)
print(matmulTensor.min())
print(matmulTensor.max())
print(matmulTensor.size())
#exercises 6-9
```
