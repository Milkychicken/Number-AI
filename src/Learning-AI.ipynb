## Tensors and Tensor operations


### Tensor basics: Initialization and Modification

```python
import torch as t
from torch import tensor

torchTensor = tensor([[1, 2, 3],
                      [4, 5, 6]])
#y = tensor([[1, 2, 3],
#            [4, 5]])
print("ndims: " + str(torchTensor.ndim) + "  " + str(torchTensor.shape))
```

A ***tensor*** is akin to an n-dimensional array of one specific datatype. Every dimension always has to be the same length, so tensor y would produce a ***ValueError*** because of invalid sizes. The tensors variable **ndim** is equal to its number of total dimensions and its variable **shape** displays its length in every one of its dimensions.

```python
print(torchTensor)
print(torchTensor.T)
```

The constant **tensor.T** is a _transposed_ (mirrored and flipped) version of the tensor.

```python
print(torchTensor.zero_())
```

```python
tensor_a = t.arange(0, 48, 4)
print(tensor_a)
```

Using **torch.arange(start, end, step),** you can create a one dimensional tensor with values from start to end that increase by step.

### Tensor operations

```python
import torch as t
torchTensor = tensor([[1, 2, 3],
                      [4, 5, 6]])
print(torchTensor)
print(torchTensor.T, end="\n\n")
print(torchTensor * torchTensor)
print(torchTensor.matmul(torchTensor.T))
```

Tensors can be **multiplied** like regular matrices, which is done by multiplying each element with the element in the same place on the other tensor.
 ***
 _Matrix multiplication_ is a special kind of multiplication that is only possible when the **inner dimensions** match (e.g. t.size(3, 2) and t.size(2, 3)). This is done by multiplying the ***dotproducts*** of the rows and columns of the tensors/matricies.

See more about matrix multiplication at: https://www.mathsisfun.com/algebra/matrix-multiplying.html


```python
tensor_a = t.arange(0, 48, 4).reshape(3, 4)
print(tensor_a)
print(tensor_a.min())
print(tensor_a.max())
#print(tensor.mean()) "TypeError: "Input dtype must be either a floating point or complex dtype. Got: Long"
print(tensor_a.mean(dtype=t.float32))
print(tensor_a.sum())
print(tensor_a.type(t.complex128)) #complex dtypes range from 32 (cHalf) to 128
```

With methods such as **min, max, mean** or **sum,** you can calculate values from the tensor. ***Be careful!*** Some methods require floating-point or complex datatypes instead of ints. This can be achieved by either specifying the type within the method or using the method .type(dtype) beforehand. For PyTorch functions and methods, **only PyTorch datatypes** can be used.

```python
print(tensor_a.argmin())
print(tensor_a.argmax())
```

The **.argmin()** and **.argmax()** methods are closely related to its min and max counterparts. Instead of the min and max value of the tensor, they both show the positions of those values **within the tensor**.
- all the calculation methods shown above can be used as static functions from the class torch as well.
- all the calculation methods shown above are called tensor aggregation

```python
print(tensor_a[2][3])
print(torchTensor[0])
```

Tensor elements can be accessed just like regular list or array elements. This is called **indexing** a tensor.

### Reshaping, viewing and squeezing a tensor

```python
import torch as t
torchTensor = t.arange(1., 13.)

print(torchTensor)
print(torchTensor.reshape(3, 4))
```

With the function **.reshape(dimension1, dimension2, dimensionN...),** an already existing tensor can be reshaped into the dimensions given. The product of all dimensions **has to be equal** to the number of elements in the tensor for it to fit the new shape.

```python
z = torchTensor.view(1, 12)
print(z)

#changing the first element of z
z[0, 0] = 500.
print(z)
print(torchTensor)
```

View is quite similar to reshape in the sense that they **both change the dimensions of a tensor.** The difference is, that reshape returns a new object, while view **shares the memory with the original element.** Therefore, even though z might look different, they both access the same element in memory so **changing one will also change the other.**

```python
x_stacked = t.stack([torchTensor, torchTensor])
print(x_stacked)
x_stacked = t.stack([torchTensor, torchTensor], dim=1)
print(x_stacked)
x_stacked = t.stack([torchTensor, torchTensor], dim=0)
print(x_stacked)
```

Stacking tensors with the **.stack(tensors[])** function will combine them to a single tensor by **concatenating/adding them on the dimension given.** Changing the parameter **dim** to another dimension will stack them in a different way. For example, with dim=1 like in the example, the tensor will become of size(12, 2), where each of the pairs of numbers consists of the numbers that were in that index **before the stack.** The default value for dim is 0.
- note: this function requires the sizes of the tensors to be equal

```python
tensor_b = t.arange(1, 11).reshape(1, 5, 1, 2)
print(f"Previous tensor: {tensor_b}")
print(f"Previous tensor shape: {tensor_b.shape}")
squeezed_tensor = tensor_b.squeeze()
print(f"New tensor: {squeezed_tensor}")
print(f"New tensor shape: {squeezed_tensor.shape}", end="\n\n")

unsqueezed_tensor = tensor_b.unsqueeze(dim=4)
print(f"New tensor: {unsqueezed_tensor}")
print(f"New tensor shape: {unsqueezed_tensor.shape}")
```

As seen in the example, tensor_b has a lot of "unnecessary" dimensions of size 1. The **.squeeze()** method simply deletes those dimensions. If you look at the size parameter, you will quickly notice all the ones being gone.
***
The **unsqueeze(dim)** method reverses this process by adding an extra dimension at the dimension given. The default value is 0 and the dimensions start from the outside in.

```python
tensor_c = t.randn(5, 2, 3)
print(f"tensor : {tensor_c}")
print(f"tensor shape : {tensor_c.shape}", end="\n\n")
permuted = tensor_c.permute(2, 0, 1)
print(f"new permuted tensor: {permuted}")
print(f"new permuted tensor shape: {permuted.shape}", end="\n\n")

#permuted = permuted.permute(3, 1, 0)       #IndexError: Dimension out of range
#permuted = permuted.permute(1, 0)          #RuntimeError: number of dimensions in the tensor input does not match the length of the desired ordering of dimensions
#print(f"new permuted tensor: {permuted}")
#print(f"new permuted tensor shape: {permuted.shape}", end="\n\n")
```

A ***Permutation*** of a tensor is similar to a *view*, as it shares the same memory as the original tensor. However, instead of choosing new sizes for the dimensions, the original dimensions are swapped around. This can be done with the **torch.permute()** function.

**Be Careful!** The passed arguments of the permute function may not exceed the index of the length of the .size() array-like attribute of the tensor and the number of arguments has to match the size the tensors size.

```python
image_tensor = t.randn(size=(64, 64, 3))    #[height, width, color channels]
print(f"Unpermuted tensor (wrong format): {image_tensor.shape}")
#reshapint => permuting:
permuted_image_tensor = image_tensor.permute(2, 0, 1)   #[color channels, height, width]
print(f"permuted tensor shape: {permuted_image_tensor.shape}", end="\n\n")
```

A common use of Permutations is in turning images into the correct data format of tensors.

### Indexing/selecting data from tensors

```python
import torch as t
torchTensor = t.arange(0, 10).reshape(1, 2, 5)
torchTensor, torchTensor.shape
```

Indexing on **pyTorch tensors** is the same as indexing on numPy arrays, which is the same as Java arrays.

```python
print(torchTensor[0])         #indexing on the first dimension
print(torchTensor[0, 0])      #indexing on the second dimension
print(torchTensor[0, 0, 0])   #indexing on the third dimension
```

```python
print(torchTensor[0])
print(torchTensor[0][0])
print(torchTensor[0][0][0])
```

Indexing works either the same as in Java, where you separate the dimensions with square brackets, or like in numPy, where you separate the dimensions with commas.
***
You can also use *":"* to select **all** of a dimension. In this example: get all elements of the 0th and 1st dimension, but only the second ones from the 2nd dimension

```python
print(torchTensor)
print(torchTensor[0, :, 1])  #0th, 1st, 2nd dimension
#selecting all from the 1st that are on index 1 in the 2nd
```

One more example: get all values from the 1st dimension that have index 1 in the 2nd dimension

```python
print(torchTensor[:, 0, 0])
print(torchTensor[0, 0, 0])
```

***Note:*** Indexing with a ":" returns a pyTorch array compared to a tensor value, because it could be multiple elements.

### PyTorch tensors & NumPy

**NumPy** is a popular Python *numerical computing library*. PyTorch actually requires NumPy to function and NumPy comes installed with it. That's why PyTorch tensors have a lot of similarities and are really compatible with NumPy ndArrays.

```python
import torch as t
import numpy as np

torchTensor = t.arange(0., 30., 5).reshape(1, 2, 3)
fromPyTorch = t.Tensor.numpy(torchTensor)
print(f"Converted to Pytorch to Numpy: {fromPyTorch}, datatype: {fromPyTorch.dtype}")
ndArray = np.arange(0., 30., 5).reshape(1, 2, 3)
fromNumPy = t.from_numpy(ndArray)
print(f"Converted from Numpy to Pytorch: {fromNumPy}, datatype: {fromNumPy.dtype}")
```

***Note:*** Be careful when converting float ndArrays to float Tensors, as tensors use float32 by default, while ndArrays use float64. This problem only occurs during the conversion from NumPy to Torch though.

```python
tTensor = t.arange(1., 6.)
ndArray = np.arange(1., 6.)
print(f"default types: {tTensor.dtype} (PyTorch), {ndArray.dtype} (NumPy)", end="\n\n")
toNp = t.Tensor.numpy(tTensor)
toTorch = t.from_numpy(ndArray)
print(f"converted types: {toNp.dtype} (to NumPy), {toTorch.dtype} (to PyTorch)")
```

### Reproducibility - taking the random out of random


In short, a **neural network** functions by generating *random numbers* and updating those through *tensor operations* to make them a better representation of what they should be. These updates can always be made the same, because you can just reuse the same learning set. The starting numbers however are generated ***pseudorandomly*** and can be controlled with setting a **random seed**.

```python
import torch as t

random_A = t.rand(3, 3)
random_B = t.rand(3, 3)
#random every time

random_A, random_B
```

```python
#Setting a seed before creating new tensors
RANDOM_SEED = 37
t.manual_seed(RANDOM_SEED)
random_C = t.rand(3, 3)
random_D = t.rand(3, 3)

print(random_C)
print(random_D)
print(random_C == random_D)
```

manual_see() ensures, that the "random" data generated is always the same, but the tensors are still different. This is because all the random numbers generated "advance the random queue". To fix this, you have to set the seed a second time before generating new random numbers of any form.

```python
#Setting a seed before creating new tensors
RANDOM_SEED = 37
t.manual_seed(RANDOM_SEED)
random_C = t.rand(3, 3)
t.manual_seed(RANDOM_SEED) #setting the seed a second time before creating a second tensor
random_D = t.rand(3, 3)

print(random_C)
print(random_D)
print(random_C == random_D)
#All true :)
```

### Running PyTorch on a GPU

PyTorch gives you the ability to choose which device you want to compute your calculations on. Thanks to NVIDIA hardware and CUDA, these computations can also be assigned to a GPU, which allows for significantly faster runtimes. Using your own GPU takes a little setup beforehand, as you need CUDA and necessary GPU drivers. CUDA can be installed with the command below:

pip3 install torch torchvision --index-url https://download.pytorch.org/whl/cu130

```python
import torch as t

print(t.cuda.is_available())
print(t.version.cuda)
print(t.cuda.device_count())
print(t.cuda.get_device_name(0))
```

To actually **use** your GPU, you have to manually tell pyTorch though. This can be done on a single **tensor** but *also* a whole ***model***.

```python
device = t.device("cuda" if t.cuda.is_available() else "cpu")

fastTensor = t.arange(1., 31.).reshape(2, 3, 5)
print(fastTensor)
print(fastTensor.device)
fastTensor = fastTensor.to(device)
print(fastTensor.device)
```

Since pyTorch can use both the GPU and the CPU, it's the **best practice** to check if cuda is available *first* and use the cpu instead if not.
